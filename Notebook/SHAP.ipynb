## SHAP Analysis – Feature Impact on Churn

SHAP (SHapley Additive exPlanations) was used to understand how the XGBoost model makes churn predictions.

### Global Feature Importance
- **Month-to-month contract** has the highest overall impact on model predictions.
- Other important features include **Tenure Months, Dependents, Monthly Charges, Contract type, Tech Support, Online Security, Total Charges, Payment Method, and Internet Service**.

### SHAP Beeswarm Analysis
- **Right side of 0** → pushes the prediction towards higher churn risk.
- **Left side of 0** → pushes the prediction towards lower churn risk.
- **Red** → high feature value/presence.
- **Blue** → low feature value/absence.
- Greater distance from 0 indicates a stronger impact on the prediction.

### Key Insights
- **Month-to-month contracts** are strongly associated with higher predicted churn risk.
- **Shorter tenure** is associated with higher churn risk, while longer-tenure customers tend to have lower predicted churn risk.
- **Higher monthly charges** tend to increase predicted churn risk.
- Customers with **no dependents** show higher predicted churn risk.
- **No Tech Support** and **No Online Security** are associated with higher predicted churn risk.
- **Electronic check** payment and **Fiber optic** internet service are also associated with higher predicted churn risk.
- A **two-year contract** tends to be associated with lower predicted churn risk.

> **Note:** SHAP explains the features that influence the model's predictions; it does not establish causation.

```python
from sklearn.model_selection import train_test_split

indices = df.index

train_idx, test_idx = train_test_split(
    indices,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print(len(test_idx))
```

```python
df_results = df.loc[test_idx].copy()

df_results["Churn_Probability"] = xgb_model.predict_proba(X_test)[:, 1]

df_results["Retention_Priority"] = (
    df_results["Churn_Probability"] * df_results["CLTV"]
)

df_results[[
    "CustomerID",
    "Churn_Probability",
    "CLTV",
    "Retention_Priority"
]].head()
```

```python
df_results["Risk_Level"] = pd.cut(
    df_results["Churn_Probability"],
    bins=[0, 0.3, 0.6, 1.0],
    labels=["Low Risk", "Medium Risk", "High Risk"],
    include_lowest=True
)

df_results["Risk_Level"].value_counts()
```

```python
df_results["Priority_Level"] = pd.qcut(
    df_results["Retention_Priority"],
    q=3,
    labels=["Low Priority", "Medium Priority", "High Priority"]
)

df_results["Priority_Level"].value_counts()
```

```python
df_results["Retention_Segment"] = "Lower Priority"

df_results.loc[
    (df_results["Risk_Level"] == "High Risk") &
    (df_results["Priority_Level"] == "High Priority"),
    "Retention_Segment"
] = "Critical"

df_results.loc[
    (df_results["Risk_Level"] == "High Risk") &
    (df_results["Priority_Level"] != "High Priority"),
    "Retention_Segment"
] = "At Risk"

df_results.loc[
    (df_results["Risk_Level"] == "Medium Risk") &
    (df_results["Priority_Level"] == "High Priority"),
    "Retention_Segment"
] = "Valuable at Risk"

df_results["Retention_Segment"].value_counts()
```

```python
segment_summary = df_results.groupby(
    "Retention_Segment",
    observed=True
).agg(
    Customers=("CustomerID", "count"),
    Total_CLTV=("CLTV", "sum"),
    Average_Churn_Probability=("Churn_Probability", "mean"),
    Total_Retention_Priority=("Retention_Priority", "sum")
).sort_values(
    "Total_Retention_Priority",
    ascending=False
)

segment_summary
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

for segment in df_results["Retention_Segment"].unique():
    subset = df_results[df_results["Retention_Segment"] == segment]
    
    plt.scatter(
        subset["Churn_Probability"],
        subset["CLTV"],
        label=segment,
        alpha=0.6
    )

plt.xlabel("Churn Probability")
plt.ylabel("Customer Lifetime Value (CLTV)")
plt.title("Customer Churn Risk × CLTV")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

## Churn Risk vs CLTV

Here we are comparing the churn probability of customers with their CLTV.

- X axis → Churn Probability
- Y axis → CLTV
- Each point represents one customer

The main idea is to find customers who have both **high churn probability and high CLTV**. These customers are more important because losing them can have a bigger impact on the company.

The upper right side of the graph represents customers who are both high risk and high value.

So, by using churn probability along with CLTV, we can decide which customers should be given more priority for retention.

**Business insight:** Just knowing who is likely to churn is not enough. We should also consider how valuable the customer is to the company.

```python
total_value_at_risk = df_results["Retention_Priority"].sum()

critical_value_at_risk = df_results.loc[
    df_results["Retention_Segment"] == "Critical",
    "Retention_Priority"
].sum()

valuable_risk_value = df_results.loc[
    df_results["Retention_Segment"] == "Valuable at Risk",
    "Retention_Priority"
].sum()

print(f"Total potential value at risk: {total_value_at_risk:,.2f}")
print(f"Critical segment value at risk: {critical_value_at_risk:,.2f}")
print(f"Valuable at Risk segment value at risk: {valuable_risk_value:,.2f}")
```

## Individual Customer SHAP Analysis

The SHAP waterfall plot shows how different features contributed to the churn prediction for a single customer.

For customer **0295-PPHDO**, the model predicted a churn probability of around **89.97%**, and the customer was classified as **Critical** based on the combination of churn probability and CLTV.

The plot starts from the average model output and then shows how each feature changes the prediction.

The main features contributing towards the prediction were:

- Tenure Months
- Month-to-month Contract
- Total Charges
- Monthly Charges
- Tech Support
- Payment Method
- Internet Service
- Dependents

Among these, **Tenure Months** and **Month-to-month Contract** had the largest individual SHAP contributions.

This helps us understand not only **which customers are likely to churn**, but also **which features are influencing the model's prediction for an individual customer**.

> SHAP values explain the model's prediction and should not be interpreted as proof that these factors directly cause customer churn.
